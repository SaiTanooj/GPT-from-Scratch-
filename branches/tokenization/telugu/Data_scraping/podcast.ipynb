```python
from youtube_transcript_api import YouTubeTranscriptApi
import json
import re
import os
import time

# ============================================================
# INPUT: ADD YOUR YOUTUBE MOVIE LINKS HERE
# ============================================================

VIDEO_URLS = [
    "https://www.youtube.com/watch?v=BuA9gmVNM3A",
    "https://www.youtube.com/watch?v=WxTQybN5_WI",
    "https://www.youtube.com/watch?v=ecNSxv90D6I",
    "https://www.youtube.com/watch?v=-gUeOaD1CBE",
    "https://www.youtube.com/watch?v=LgjEP0dsQZc",
    "https://www.youtube.com/watch?v=CPIFWca5NaU",
    "https://www.youtube.com/watch?v=k2SKajAwsfM",
    "https://www.youtube.com/watch?v=Z907ENCogm4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=6",
    "https://www.youtube.com/watch?v=n4HC63jy0EI&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=7",
    "https://www.youtube.com/watch?v=y58ppLTeIk8&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=8",
    "https://www.youtube.com/watch?v=CAugJNXWOAA&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=9",
    "https://www.youtube.com/watch?v=34ohxu64NZk&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=10",
    "https://www.youtube.com/watch?v=Sn-FIwkb5Pw&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=11",
    "https://www.youtube.com/watch?v=hK47JsHu7kg&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=13",
    "https://www.youtube.com/watch?v=MEryjjZwKb4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=14",
    "https://www.youtube.com/watch?v=OtcyBONn_8I&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=19",
    "https://www.youtube.com/watch?v=qHaQA8CG1VM&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=20",
    "https://www.youtube.com/watch?v=OSjir7mDCFs&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=21",
    "https://www.youtube.com/watch?v=FS1DI5D_hUk&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=22",
    "https://www.youtube.com/watch?v=JzKTnvF0OgU&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=23",
    "https://www.youtube.com/watch?v=xRNQp-XuJuU&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=24",
    "https://www.youtube.com/watch?v=a_DtVXENkAg&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=26",
    "https://www.youtube.com/watch?v=pK-yGAs3bMI&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=27",
    "https://www.youtube.com/watch?v=wAp3ROyREmE&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=28",
    "https://www.youtube.com/watch?v=q4TOWkhqFTA&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=30",
    "https://www.youtube.com/watch?v=VlOvr8tp5ao&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=31",
    "https://www.youtube.com/watch?v=pM0PSg8u3nY&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=32",
    "https://www.youtube.com/watch?v=73CByp9O6kc&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=34",
    "https://www.youtube.com/watch?v=Oq8XZLi_c58&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=35",
    "https://www.youtube.com/watch?v=QhqEV78uj58&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=37",
    "https://www.youtube.com/watch?v=2ADSAW_D_P8&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=38",
    "https://www.youtube.com/watch?v=5diz3ZtVT5U&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=39",
    "https://www.youtube.com/watch?v=mkPWFBmCpw8&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=40",
    "https://www.youtube.com/watch?v=uKQIRFr0oYg&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=42",
    "https://www.youtube.com/watch?v=U4l2vgUsBOI&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=43",
    "https://www.youtube.com/watch?v=5GhXx_a9CBM&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=44",
    "https://www.youtube.com/watch?v=TLmYbVXX7MU&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=45",
    "https://www.youtube.com/watch?v=SO25vaOLT2k&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=46",
    "https://www.youtube.com/watch?v=KiGH8d1Ww-c&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=47",
    "https://www.youtube.com/watch?v=UUPiLFA-Og4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=50",
    "https://www.youtube.com/watch?v=7FbeLGB6lr0&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=51",
    "https://www.youtube.com/watch?v=ycNyo7rpaMA&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=52",
    "https://www.youtube.com/watch?v=O-Y01hFdFM4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=55",
    "https://www.youtube.com/watch?v=CjsldB37dHU&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=56",
    "https://www.youtube.com/watch?v=CaUUFXANhHE&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=58",
    "https://www.youtube.com/watch?v=YgQRQXQA8KY&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=59",
    "https://www.youtube.com/watch?v=ShwnlRlUKA4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=61",
    "https://www.youtube.com/watch?v=eK71EN3P78U&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=62",
    "https://www.youtube.com/watch?v=jyDav--HeZI&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=63",
    "https://www.youtube.com/watch?v=KKLsBurqO2E&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=64",
    "https://www.youtube.com/watch?v=lDF6KLHKrUs&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=66",
    "https://www.youtube.com/watch?v=4cfhcFl8Kxs&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=68",
    "https://www.youtube.com/watch?v=s-jbL-3jupA&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=73",
    "https://www.youtube.com/watch?v=_HK5Jge3hO4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=79",
    "https://www.youtube.com/watch?v=7QNI51lFjlw&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=85",
    "https://www.youtube.com/watch?v=XZqK-I-kc-Y&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=92",
    "https://www.youtube.com/watch?v=ra1__kEJMVo&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=96",
    "https://www.youtube.com/watch?v=OnAnJgv0AEY&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=101"
    "https://www.youtube.com/watch?v=sVf1xs97yPo&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=102",
    "https://www.youtube.com/watch?v=0Fs_Zh4fG3E&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=104",
    "https://www.youtube.com/watch?v=S4IsJNcWCy4&list=PLKdv9M1lZ40csEVtFvIzWn8Yp3-nfXiWe&index=140",
    ""
    
    

    # Add more movies here
    # "https://www.youtube.com/watch?v=XXXXXXXXXXX",
    # "https://www.youtube.com/watch?v=YYYYYYYYYYY",
    # "https://www.youtube.com/watch?v=ZZZZZZZZZZZ",
]


# ============================================================
# OUTPUT FILE
# ============================================================

OUTPUT_FILE = "telugu_podcast.json"


# ============================================================
# YOUTUBE TRANSCRIPT API
# ============================================================

api = YouTubeTranscriptApi()


# ============================================================
# EXTRACT VIDEO ID
# ============================================================

def get_video_id(url):

    # Normal YouTube URL
    match = re.search(
        r"(?:youtube\.com/watch\?v=|youtu\.be/)([^&?/]+)",
        url
    )

    if match:
        return match.group(1)

    return None


# ============================================================
# EXTRACT TELUGU SUBTITLES
# ============================================================

def extract_telugu_subtitles(video_id):

    try:

        print("Checking subtitles...")

        # Get available transcripts
        transcript_list = api.list(video_id)

        print("Available subtitle tracks:")

        found_telugu = False

        for transcript in transcript_list:

            print(
                f"   {transcript.language} "
                f"({transcript.language_code})"
            )

            if transcript.language_code == "te":

                found_telugu = True

        if not found_telugu:

            print("   Telugu subtitle track NOT found.")

            return None

        # ----------------------------------------------------
        # Fetch Telugu transcript
        # ----------------------------------------------------

        print("Fetching Telugu subtitles...")

        transcript = api.fetch(
            video_id,
            languages=["te"]
        )

        # ----------------------------------------------------
        # Combine subtitle segments
        # ----------------------------------------------------

        text_parts = []

        for item in transcript:

            text_parts.append(item.text)

        text = " ".join(text_parts)

        # ----------------------------------------------------
        # Clean excessive whitespace
        # ----------------------------------------------------

        text = re.sub(
            r"\s+",
            " ",
            text
        ).strip()

        return text

    except Exception as e:

        print(
            f"ERROR: {type(e).__name__}: {e}"
        )

        return None


# ============================================================
# PROCESS ALL MOVIES
# ============================================================

movies = []

total_movies = len(VIDEO_URLS)

print("=" * 70)
print(f"TOTAL MOVIES TO PROCESS: {total_movies}")
print("=" * 70)


for index, url in enumerate(
    VIDEO_URLS,
    start=1
):

    print("\n")
    print("=" * 70)

    print(
        f"MOVIE {index}/{total_movies}"
    )

    print("=" * 70)

    print("URL:")
    print(url)

    # --------------------------------------------------------
    # Get video ID
    # --------------------------------------------------------

    video_id = get_video_id(url)

    if not video_id:

        print("ERROR: Could not extract video ID.")

        continue

    print(
        f"Video ID: {video_id}"
    )

    # --------------------------------------------------------
    # Extract subtitles
    # --------------------------------------------------------

    text = extract_telugu_subtitles(
        video_id
    )

    if not text:

        print(
            "[SKIPPED] No Telugu subtitles."
        )

        continue

    # --------------------------------------------------------
    # Word count
    # --------------------------------------------------------

    word_count = len(
        text.split()
    )

    character_count = len(text)

    # --------------------------------------------------------
    # Create movie record
    # --------------------------------------------------------

    movie_data = {
        "source":"yt",

        "id": len(movies) + 1,

        "video_id": video_id,

        "url": url,

        "language": "te",

        "text": text,

        "word_count": word_count,

        "character_count": character_count
    }

    movies.append(
        movie_data
    )

    print(
        f"[SAVED] Movie {len(movies)}"
    )

    print(
        f"Characters: {character_count:,}"
    )

    print(
        f"Words: {word_count:,}"
    )

    # Small delay between videos
    time.sleep(1)


# ============================================================
# SAVE ALL MOVIES TO ONE JSON FILE
# ============================================================

output_data = {

    "language": "te",

    "total_movies": len(movies),

    "total_words": sum(
        movie["word_count"]
        for movie in movies
    ),

    "total_characters": sum(
        movie["character_count"]
        for movie in movies
    ),

    "movies": movies
}


with open(
    OUTPUT_FILE,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        output_data,
        f,
        ensure_ascii=False,
        indent=2
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("EXTRACTION COMPLETE")
print("=" * 70)

print(
    f"Movies requested : {total_movies}"
)

print(
    f"Movies extracted : {len(movies)}"
)

print(
    f"Total words      : "
    f"{output_data['total_words']:,}"
)

print(
    f"Total characters : "
    f"{output_data['total_characters']:,}"
)

print(
    f"Output file      : {OUTPUT_FILE}"
)

print("=" * 70)
```
