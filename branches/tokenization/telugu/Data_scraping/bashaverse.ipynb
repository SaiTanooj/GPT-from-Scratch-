```python
import pandas as pd
import json
import os

from huggingface_hub import hf_hub_download, login


# ============================================================
# 3. HUGGING FACE TOKEN
# ============================================================
#
# Paste your Hugging Face token between the quotes.
#
# Example:
# HF_TOKEN = "hf_xxxxxxxxxxxxxxxxxxxxx"
#

HF_TOKEN = "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"


# ============================================================
# 4. LOGIN TO HUGGING FACE
# ============================================================

if HF_TOKEN.strip() == "":
    raise ValueError(
        "Please enter your Hugging Face token in HF_TOKEN before running."
    )

login(token=HF_TOKEN)

print("Hugging Face authentication successful!")


# ============================================================
# 5. DATASET CONFIGURATION
# ============================================================

REPO_ID = "ltrciiith/bhashik-parallel-corpora-generic"

# Start and end shard
START_SHARD = 1
END_SHARD = 20

# Output JSON file
OUTPUT_FILE = "/kaggle/working/bhashaverse_telugu.json"


# ============================================================
# 6. STORAGE
# ============================================================

# Final JSON records
all_data = []

# Used to remove duplicate Telugu sentences
telugu_sentences = set()


# ============================================================
# 7. PROCESS PARQUET FILES ONE BY ONE
# ============================================================

for shard_number in range(START_SHARD, END_SHARD + 1):

    filename = f"data/train-shard_{shard_number:08d}.parquet"

    print("\n" + "=" * 70)
    print(f"PROCESSING SHARD {shard_number}")
    print(filename)
    print("=" * 70)

    try:

        # ----------------------------------------------------
        # Download ONLY this Parquet file
        # ----------------------------------------------------

        file_path = hf_hub_download(
            repo_id=REPO_ID,
            filename=filename,
            repo_type="dataset",
            token=HF_TOKEN
        )

        print("Parquet downloaded successfully.")
        print("Path:", file_path)


        # ----------------------------------------------------
        # Read Parquet into DataFrame
        # ----------------------------------------------------

        df = pd.read_parquet(file_path)

        print("DataFrame shape:", df.shape)


        # ====================================================
        # EXTRACT TELUGU SOURCE TEXT
        # ====================================================

        source_rows = df[
            df["source_language"] == "tel_Telu"
        ]

        source_texts = (
            source_rows["source_text"]
            .dropna()
        )

        source_count = 0

        for text in source_texts:

            text = str(text).strip()

            if text and text not in telugu_sentences:

                telugu_sentences.add(text)

                all_data.append({
                    "source": "bhashaverse",
                    "url": filename,
                    "text": text
                })

                source_count += 1


        # ====================================================
        # EXTRACT TELUGU TARGET TEXT
        # ====================================================

        target_rows = df[
            df["target_language"] == "tel_Telu"
        ]

        target_texts = (
            target_rows["target_text"]
            .dropna()
        )

        target_count = 0

        for text in target_texts:

            text = str(text).strip()

            if text and text not in telugu_sentences:

                telugu_sentences.add(text)

                all_data.append({
                    "source": "bhashaverse",
                    "url": filename,
                    "text": text
                })

                target_count += 1


        # ====================================================
        # SHARD STATISTICS
        # ====================================================

        print("\nShard completed.")

        print(
            "Telugu source sentences:",
            source_count
        )

        print(
            "Telugu target sentences:",
            target_count
        )

        print(
            "Total unique Telugu sentences:",
            len(telugu_sentences)
        )


        # ====================================================
        # SAVE JSON AFTER EVERY SHARD
        # ====================================================

        with open(
            OUTPUT_FILE,
            "w",
            encoding="utf-8"
        ) as f:

            json.dump(
                all_data,
                f,
                ensure_ascii=False,
                indent=2
            )

        print("\nJSON updated successfully.")


        # ====================================================
        # REMOVE DOWNLOADED PARQUET
        # ====================================================
        #
        # This saves Kaggle disk space.
        #

        if os.path.exists(file_path):

            os.remove(file_path)

            print(
                "Downloaded Parquet removed from disk."
            )


        # Remove DataFrame from memory
        del df


    except Exception as e:

        print("\nERROR PROCESSING SHARD:", shard_number)
        print(e)

        print(
            "Skipping this shard and moving to the next one."
        )

        continue


# ============================================================
# FINAL RESULT
# ============================================================

print("\n")
print("=" * 70)
print("EXTRACTION COMPLETE")
print("=" * 70)

print(
    "Total unique Telugu sentences:",
    len(telugu_sentences)
)

print(
    "Total JSON records:",
    len(all_data)
)

print(
    "JSON saved at:",
    OUTPUT_FILE
)
```
