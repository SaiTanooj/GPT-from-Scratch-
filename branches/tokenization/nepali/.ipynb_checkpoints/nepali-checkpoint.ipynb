### Text cleaning 
- Removed HTML tags and URLs from the text.
- Normalized the Nepali text using Unicode NFC normalization.
- Kept only Nepali characters, spaces, numbers, and basic punctuation.
- Removed unwanted characters and extra tabs/spaces.
- Normalized multiple newlines into a single newline.
- Added spaces around punctuation marks for consistent tokenization.
- Removed unnecessary leading and trailing spaces.

```python
import re, unicodedata

html_re = re.compile(r'<[^>]*>')
url_re = re.compile(r'http\S+|www\.\S+', re.IGNORECASE)
keep_re = re.compile(r'[^\u0900-\u097F\u200D\u200C\s.,!?0-9]')
punct_re = re.compile(r'([.,!?।॥])')
ws_re = re.compile(r'\s+')
nl_re = re.compile(r'\n+')
tab_re = re.compile(r'[ \t]+')

def clean_text(text):
    text = html_re.sub('', text)
    text = url_re.sub('', text)
    text = unicodedata.normalize('NFC', text)
    text = keep_re.sub('', text)
    text = tab_re.sub(' ', text)
    text = nl_re.sub('\n', text).strip()
    text = punct_re.sub(r' \1 ', text)
    text = tab_re.sub(' ', text).strip()
    return text
```

```python
for d in [df1, df2, df3, df4, df5]:
    d["text"] = d["text"].map(clean_text)
```

### Adding the dataframes to the buckets
- Created 256 bucket files to store the Nepali sentences separately.
- Read the text from all five datasets and Split the text into individual lines/sentences using newline characters.
- Generated an MD5 hash for each sentence to determine its bucket and Stored each sentence in one of the 256 buckets based on its hash.
- Added the prefix `M` to identify sentences coming from the manual datasets.
- Counted the total number of manual sentence candidates written to disk.


```python
import hashlib
import os
import glob
import shutil

BUCKET_DIR = "/kaggle/working/nepali_buckets"
N_BUCKETS = 256

# Start clean if notebook is being rerun
if os.path.exists(BUCKET_DIR):
    shutil.rmtree(BUCKET_DIR)

os.makedirs(BUCKET_DIR)

# Open bucket files
manual_files = [
    open(
        os.path.join(BUCKET_DIR, f"bucket_{i:03d}.txt"),
        "w",
        encoding="utf-8"
    )
    for i in range(N_BUCKETS)
]

manual_candidates = 0

for d in [df1, df2, df3, df4, df5]:

    for text in d["text"].values:

        for line in text.split("\n"):

            line = line.strip()

            if not line:
                continue

            h = int(
                hashlib.md5(
                    line.encode("utf-8")
                ).hexdigest()[:2],
                16
            )

            manual_files[h].write(
                "M\t" + line + "\n"
            )

            manual_candidates += 1

for f in manual_files:
    f.close()

print(
    manual_candidates,
    "manual sentence candidates written to disk"
)
```

### Reading the large nepali data corpus 
- Streamed the Nepali corpus directly from Hugging Face without loading the entire dataset into memory.
- Read the JSON file record by record using `ijson`.
- Extracted the `text` field from each record and skipped empty records.
- Collected approximately 8 GB of text from the downloaded corpus and cleaned the text 
- Generated an MD5 hash for each sentence to assign it to one of the 256 buckets.
- Stored each downloaded sentence in its corresponding bucket with added the tag D to each sentence .


```python
import requests
import ijson

url = "https://huggingface.co/datasets/Tantade/Nepalicorpus/resolve/main/nepali_merged_all.json"

TARGET_BYTES = 8* 1024**3

downloaded_files = [
    open(
        os.path.join(BUCKET_DIR, f"bucket_{i:03d}.txt"),
        "a",
        encoding="utf-8"
    )
    for i in range(N_BUCKETS)
]

total_bytes = 0
total_records = 0

with requests.get(url, stream=True) as r:
    r.raise_for_status()

    for i, obj in enumerate(ijson.items(r.raw, "item"), 1):

        text = obj.get("text")

        if not text:
            continue

        total_bytes += len(text.encode("utf-8"))

        text = clean_text(text)

        for line in re.split(r"[।॥\n]", text):
            line = line.strip()

            if not line:
                continue

            bucket = int(
                hashlib.md5(
                    line.encode("utf-8")
                ).hexdigest()[:2],
                16
            )

            downloaded_files[bucket].write(
                "D\t" + line + "\n"
            )

        total_records += 1

        if i % 500000 == 0:
            print(
                f"{i} records, "
                f"{total_bytes / 1024**3:.2f} GB collected"
            )

        if total_bytes >= TARGET_BYTES:
            print(
                f"Reached 11 GB at {i} records"
            )
            break

for f in downloaded_files:
    f.close()

print("Downloaded records:", total_records)
```

#### Deduplication phase 
- Created separate files for the train, val, and test.
- Processed one bucket at a time Checked each sentence for duplicates within its bucket using a set.
- Counted the total number of unique and duplicate sentences.
- Kept track of unique sentences from manual and downloaded sources separately.
- Counted the number of words from manual and downloaded sentences.
- Used a deterministic hash-based split to divide the data into 80% training, 10% validation, and 10% test sets.
- Wrote each sentence to its corresponding split file.


```python
TRAIN_PATH = "/kaggle/working/nepali_train.txt"
VAL_PATH = "/kaggle/working/nepali_val.txt"
TEST_PATH = "/kaggle/working/nepali_test.txt"

# Open final split files now
train_out = open(
    TRAIN_PATH,
    "w",
    encoding="utf-8"
)

val_out = open(
    VAL_PATH,
    "w",
    encoding="utf-8"
)

test_out = open(
    TEST_PATH,
    "w",
    encoding="utf-8"
)

unique_total = 0
duplicate_total = 0

manual_unique = 0
downloaded_unique = 0

manual_words = 0
downloaded_words = 0

# Process ONE bucket at a time
for b in range(N_BUCKETS):

    bucket_path = os.path.join(
        BUCKET_DIR,
        f"bucket_{b:03d}.txt"
    )

    seen = {}

    with open(
        bucket_path,
        "r",
        encoding="utf-8"
    ) as f:

        for raw in f:

            source, line = raw.rstrip("\n").split(
                "\t",
                1
            )

            # Duplicate sentence
            if line in seen:

                duplicate_total += 1
                continue

            seen[line] = source

            unique_total += 1

            if source == "M":

                manual_unique += 1
                manual_words += len(line.split())

            else:

                downloaded_unique += 1
                downloaded_words += len(line.split())

            # Deterministic 80/10/10 split
            split_value = (
                int(
                    hashlib.md5(
                        line.encode("utf-8")
                    ).hexdigest(),
                    16
                ) % 100
            )

            if split_value < 80:

                train_out.write(
                    line + "\n"
                )

            elif split_value < 90:

                val_out.write(
                    line + "\n"
                )

            else:

                test_out.write(
                    line + "\n"
                )

    # Delete bucket after processing
    os.remove(bucket_path)

    del seen

    if (b + 1) % 16 == 0:

        print(
            f"Processed {b + 1}/{N_BUCKETS} buckets | "
            f"unique={unique_total:,}"
        )

train_out.close()
val_out.close()
test_out.close()

print("\nDeduplication completed.")

print(
    "Unique total:",
    f"{unique_total:,}"
)

print(
    "Duplicates removed:",
    f"{duplicate_total:,}"
)

print(
    "Manual unique:",
    f"{manual_unique:,}"
)

print(
    "Downloaded unique:",
    f"{downloaded_unique:,}"
)

print(
    "Manual word count:",
    f"{manual_words:,}"
)

print(
    "Downloaded word count:",
    f"{downloaded_words:,}"
)
```

#### Train,test and validation split count and ratio

```python
import gc

gc.collect()

def count_lines(path):

    with open(
        path,
        encoding="utf-8"
    ) as f:

        return sum(
            1 for line in f
            if line.strip()
        )

n_train = count_lines(TRAIN_PATH)
n_val = count_lines(VAL_PATH)
n_test = count_lines(TEST_PATH)

n = n_train + n_val + n_test

print(
    "Train:",
    n_train,
    f"({n_train/n*100:.2f}%)"
)

print(
    "Validation:",
    n_val,
    f"({n_val/n*100:.2f}%)"
)

print(
    "Test:",
    n_test,
    f"({n_test/n*100:.2f}%)"
)

print(
    "Total:",
    n
)
```

#### Training the sentence piece tokenizer using 5k and 10k vocab splits 

```python
with open(TRAIN_PATH, encoding="utf-8") as f:
    n_lines = sum(1 for _ in f)
print(n_lines, "lines in nepali_train.txt")

vocab_sizes = [5000, 10000]
model_type = "unigram"

for v in vocab_sizes:
    print(f"\ntraining model_type={model_type}, vocab_size={v}...")
    spm.SentencePieceTrainer.train(
        input=TRAIN_PATH,
        model_prefix=f"nepali_sp_{model_type}_{v}",
        vocab_size=v,
        model_type=model_type,
        character_coverage=1.0,
        normalization_rule_name="identity",
        byte_fallback=True,
        pad_id=0,
        unk_id=1,
        bos_id=2,
        eos_id=3,
        pad_piece="<pad>",
        unk_piece="<unk>",
        bos_piece="<s>",
        eos_piece="</s>",
        shuffle_input_sentence=True,
        input_sentence_size=200000,
        train_extremely_large_corpus=True,
        num_threads=2
    )
    print(f"done: model_type={model_type}, vocab_size={v}")
```

### Testing on validation split 
- Loaded up to 5,000 validation sentences for tokenizer evaluation.
- Tokenized the validation sentences using each trained SentencePiece model.
- Calculated the average number of tokens per word (fertility).
- Calculated the percentage of unknown (`UNK`) tokens.
- Tested different vocabulary sizes using the Unigram tokenizer model.
- Compared the tokenizer models based on their UNK rate and fertility.
- Selected the model with the lowest UNK rate.
- Saved the selected SentencePiece model and vocabulary as the final tokenizer.

```python
import shutil

MAX_VAL_LINES = 5000

def load_validation_sample(path, n=5000):
    texts = []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            texts.append(line)
            if len(texts) >= n:
                break
    return texts

val_texts = load_validation_sample(VAL_PATH, MAX_VAL_LINES)
print("Validation lines used:", len(val_texts))

def quick_stats(sp, texts, n=5000, print_every=1000):
    n_tok = 0
    n_word = 0
    n_unk = 0
    unk_id = sp.unk_id()
    for i, line in enumerate(texts[:n], 1):
        ids = sp.encode(line, out_type=int)
        n_tok += len(ids)
        n_word += len(line.split())
        n_unk += sum(1 for t in ids if t == unk_id)
        if i % print_every == 0:
            print(" ", i, "lines tokenized")
    return {
        "fertility": n_tok / max(n_word, 1),
        "unk_rate": n_unk / max(n_tok, 1)
    }

model_type = "unigram"
results = {}
for v in vocab_sizes:
    key = (model_type, v)
    print(f"\nmodel_type={model_type}, vocab_size={v}")
    sp_v = spm.SentencePieceProcessor(
        model_file=f"nepali_sp_{model_type}_{v}.model"
    )
    results[key] = quick_stats(sp_v, val_texts)
    print(" ", results[key])

# Keep your existing selection logic:
# lowest UNK rate, with fertility shown for comparison.
best_key = min(
    results,
    key=lambda k: (results[k]["unk_rate"], results[k]["fertility"])
)
best_type, best_vocab = best_key
print(f"\npicking model_type={best_type}, vocab_size={best_vocab} (lowest UNK rate)")
shutil.copy(f"nepali_sp_{best_type}_{best_vocab}.model", "nepali_sp.model")
shutil.copy(f"nepali_sp_{best_type}_{best_vocab}.vocab", "nepali_sp.vocab")
sp = spm.SentencePieceProcessor(model_file="nepali_sp.model")
```

#### Saving the tokenizer and the vocab file of the best model 

```python
import os, shutil, json

out_root = (
    "/kaggle/working"
    if os.path.exists("/kaggle/working")
    else "."
)

out_dir = os.path.join(
    out_root,
    "nepali/tokenizer"
)

os.makedirs(
    out_dir,
    exist_ok=True
)

for fname in [
    "nepali_sp.model",
    "nepali_sp.vocab"
]:

    shutil.copy(
        fname,
        os.path.join(
            out_dir,
            fname
        )
    )

vocab = [
    {
        "id": i,
        "piece": sp.id_to_piece(i),
        "score": sp.get_score(i)
    }

    for i in range(
        sp.vocab_size()
    )
]

with open(
    os.path.join(
        out_dir,
        "vocab.json"
    ),
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        vocab,
        f,
        ensure_ascii=False,
        indent=2
    )

print(
    "saved to",
    out_dir
)

print(
    os.listdir(out_dir)
)
```

#### Used the best model to genrste tokens on the train data 

```python
import numpy as np
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file="/kaggle/working/nepali/tokenizer/nepali_sp.model")
TRAIN_PATH = "nepali_train.txt"
OUT_PATH = "nepali_train.bin"
CHUNK_SIZE = 500_000  # lines per chunk

vocab_size = sp.get_piece_size()
dtype = np.uint16 if vocab_size < 65536 else np.uint32

total_tokens = 0
total_lines = 0
chunk = []

with open(TRAIN_PATH, encoding="utf-8") as f_in, open(OUT_PATH, "wb") as f_out:
    for line in f_in:
        line = line.strip()
        if not line:
            continue
        ids = sp.encode(line, out_type=int)
        chunk.extend(ids)
        total_tokens += len(ids)
        total_lines += 1

        if total_lines % CHUNK_SIZE == 0:
            np.array(chunk, dtype=dtype).tofile(f_out)
            chunk = []
            print(f"  {total_lines:,} lines processed, {total_tokens:,} tokens so far")

    if chunk:  # flush remainder
        np.array(chunk, dtype=dtype).tofile(f_out)

print(f"\nFinal: {total_lines:,} lines, {total_tokens:,} tokens in {TRAIN_PATH}")
print(f"Saved to {OUT_PATH}, dtype={dtype.__name__}")
```
